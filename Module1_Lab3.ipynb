### In Chapter 1.9.4 Shah describes Data Analytics to being prone to bias. When a dataset is analyzed to get information, that information is a direct result of the dataset used. "The output of the data analytics is limited to the details of the input." This is because, if a dataset is skewed to a particular outcome, it's not the most accurate representation of the whole, so it could be a mistake to say that the best average is the best choice. The max rating is not the best way to see which company provides the best service, and the min rating tells about the most unhappy customer.  The risk involved is that a very low rating can pull down the average, and this could be a one time occurance. This is why variance helps to tell if the average is a dependable method to choose the best company.  If the existing data structure contains bias, by establishing the bias, you can prevent it's amplification.

```python
import pandas as pd

# our dictionary
data = { 'Geico': [4.7, 8.3, 9.2], 'Progressive':[7.4, 6.7, 8.9], 'USAA':[3.8, 6.3, 8.1] }
df = pd.DataFrame(data)
print(df)
```

#### Pandas lets us visualize our data above.

```python
geico_ratings = [4.7, 8.3, 9.2]
progressive_ratings = [7.4, 6.7, 8.9]
usaa_ratings = [3.8, 6.3, 8.1]
```

#### These three lists above can be used with Numpy to find the ***max, min, mean, median, variance, and standard deviation.***  The next 3 code blocks will print these statistics.

```python
import numpy as np 
geico_max = np.max(geico_ratings)
geico_min = np.min(geico_ratings)
geico_mean= np.mean(geico_ratings)
geico_variance = np.var(geico_ratings)
geico_standarddev = np.std(geico_ratings)
geico_median = np.median(geico_ratings)

print("geico max",geico_max)
print("geico min",geico_min)
print("geico mean {:4f}".format( geico_mean))
print("geico variance {:4f}".format( geico_variance))
print("geico std {:4f}".format(geico_standarddev))
print("geico median {:4f}".format(geico_median))
```

```python
progressive_max = np.max(progressive_ratings)
progressive_min = np.min(progressive_ratings)
progressive_mean= np.mean(progressive_ratings)
progressive_variance = np.var(progressive_ratings)
progressive_standarddev = np.std(progressive_ratings)
progressive_median = np.median(progressive_ratings)
print("progressive max",progressive_max)
print("progressive min",progressive_min)
print("progressive mean {:4f}".format( progressive_mean))
print("progressive variance {:4f}".format( progressive_variance))
print("progressive std {:4f}".format(progressive_standarddev))
print("progressive median {:4f}".format(progressive_median))
```

```python
usaa_max = np.max(usaa_ratings)
usaa_min = np.min(usaa_ratings)
usaa_mean= np.mean(usaa_ratings)
usaa_variance = np.var(usaa_ratings)
usaa_standarddev = np.std(usaa_ratings)
usaa_median = np.median(usaa_ratings)
print("usaa max",usaa_max)
print("usaa min",usaa_min)
print("usaa mean {:4f}".format( usaa_mean))
print("usaa variance {:4f}".format( usaa_variance))
print("usaa std {:4f}".format(usaa_standarddev))
print("usaa median {:4f}".format(usaa_median))
```

### So which is better, Progressive or Geico? To address fairness and address biases, we make sure that the data has been cleaned, Each company was allowed only three data points, which is a clean dataset.  A data scientist may say that data science is about working with increasingly large datasets in the billions, and this allows models to generalize better.  For this example we can look at the standard deviation and the variance between these three points to choose a company. The one with the lowest variance or spread will tell us there is less fluctuations of service and will be the best. Wikipedia says "Standard deviation is found by taking the square root of the average of the squared deviations of the values subtracted from their average value."  By comparing a low variance with a high mean we could decide which company will be best.

```python
#Standard deviation: find the average, subtract the average from each point, square each difference, add them together, divide by the number of points, then take the square root.
insurance_variances = [geico_variance, progressive_variance, usaa_variance]
insurance_standarddev = [geico_standarddev, progressive_standarddev, usaa_standarddev]

min_variance = np.min(insurance_variances)
min_standarddev = np.min(insurance_standarddev)

print("Lowest Variance: ",min_variance)
print("Lowest Standard Deviation: ",min_standarddev)

insurance_variance_key = {'Geico':[geico_variance], 'Progessive':[progressive_variance], 'USAA':[usaa_variance]}
print("-"*40)
insurance_standarddev_key = {'Geico':[geico_standarddev], 'Progessive':[progressive_standarddev], 'USAA':[usaa_standarddev]}

df_3 = pd.DataFrame(insurance_variance_key)
print("[Variance]")
print(df_3)
print("-"*40)
df_4 = pd.DataFrame(insurance_standarddev_key)
print("[Standard Deviation]")
print(df_4)
```
